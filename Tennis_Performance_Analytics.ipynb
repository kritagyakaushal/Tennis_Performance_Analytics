# Tennis Performance Analytics

## Project Objective
The aim of this project is to use Python to analyze tennis match statistics and visualize performance.

## Questions I Want to Answer
1. What percentage of first serves go in?
2. How many double faults occur in each match?
3. How do winners vary between matches?

## Tools
- Python
- Pandas
- Matplotlib
- Google Colab

## Dataset
I will begin with sample practice data to test my code. These records are not real match results.

```python


import pandas as pd
import matplotlib.pyplot as plt

# Practice data only — these are not real match results
data = {
    "Match": ["Match 1", "Match 2", "Match 3", "Match 4"],
    "First_Serves_In": [32, 38, 29, 41],
    "Serve_Attempts": [50, 55, 48, 56],
    "Double_Faults": [3, 2, 5, 1],
    "Winners": [12, 16, 9, 18]
}

df = pd.DataFrame(data)

print(df)
```

```python

df["First_Serve_Percent"] = (
    df["First_Serves_In"] / df["Serve_Attempts"] * 100
).round(1)

print(df[["Match", "First_Serve_Percent"]])
```

```python

df.plot(
    x="Match",
    y="First_Serve_Percent",
    kind="bar",
    legend=False,
    title="First-Serve Percentage by Match"
)

plt.ylabel("First serves in (%)")
plt.ylim(0, 100)
plt.tight_layout()
plt.show()
```

```python

df.plot(
    x="Match",
    y=["Winners", "Double_Faults"],
    kind="bar",
    title="Winners vs Double Faults by Match"
)

plt.ylabel("Number of shots")
plt.tight_layout()
plt.show()
```

## Analysis and Observations

1. Match 4 had the highest number of winners, with 18.
2. Match 3 had the highest number of double faults, with 5.
3. These statistics describe different aspects of performance. More information, such as points won and match results, would be needed to judge overall performance.

Note: The dataset contains sample practice data, not real match results.

```python

df.plot(
    x="Match",
    y="First_Serve_Percent",
    kind="line",
    marker="o",
    legend=False,
    title="First-Serve Percentage Across Matches"
)

plt.ylabel("First-serve percentage (%)")
plt.ylim(0, 100)
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Summary of Findings

- Match 4 had the highest first-serve percentage, at approximately 73.2%.
- Match 4 also had the most winners, with 18.
- Match 3 had the most double faults, with 5.
